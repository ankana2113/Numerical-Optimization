# Jacobi Method

```python
import numpy as np
```

```python
def jacobi(A,b, x_0, tol, K):
    x_iter = []
    n = A.shape[0]
    err = tol + 1
    x = x_0.copy()
    k = 0
    while err > tol and k < K:
        x_old = x.copy()
        x = jacobiIter(A,b,x_old)
        x_iter.append(x)
        k = k + 1
        err = np.linalg.norm(x-x_old,np.inf)/np.linalg.norm(x,np.inf)
        # err = np.linalg.norm(x-x_old,2)/np.linalg.norm(x,2)
    print(k)
    return x, x_iter


def jacobiIter(A,b,xold):
    n = A.shape[0]
    x = np.zeros_like(xold)
    for i in range(n):
        s = 0
        for j in range(n):
            if j!=i:
                s += A[i,j]*xold[j]
        x[i] = (b[i]- s)/A[i,i]
    return x
```

```python
A = np. array([[2., 1., 0., 0.],
               [2., 3.,  0., 1.],
               [0., -1., -6., 2.],
               [0., 2., 1., -4.]])

b = np. array([1., 0., -2., -5 ]) 

tol = 0.0001
K = 1000
x_0 = np.zeros(4)
x, x_iter = jacobi(A,b, x_0, tol, K)
print(x)
len(x_iter)
```

```python
def error(x_iter):
    errors = []
    for i in range(len(x_iter)):
        err = np.linalg.norm(A@x_iter[i] - b)
        errors.append(err)
    return errors
```

# Compare with Gaussian Elimination with Partial Pivoting

```python
def gaussian_partial(A, b):
    n = A.shape[0]
    C=np.c_[A,b.reshape(-1,1)]
    
    flag = 0
    
    # column
    for i in range(n-1):
            
        max_c, chosen_k  = 0, i
        
        # find pivot
        for k in range(i, n):
            if np.abs(C[k,i]) > max_c:
                max_c = np.abs(C[k,i])
                chosen_k = k
         
        # check if the pivot=0
        if max_c == 0:
            flag = 1
            break
        
        # interchange rows
        if chosen_k != i:
            # swap 2 rows
            temp = C[i,:].copy()
            C[i,:] = C[chosen_k,:]
            C[chosen_k,:] = temp
    
        
        # row
        for j in range(i+1, n):
            c = C[j,i]/C[i,i]
            C[j,:] = C[j,:] - c*C[i,:]
            
    return C, flag
        

def backsubstitution(T):
    flag=0
    n = T.shape[0]
    X = np.zeros((n))
    if T[n-1,n-1] == 0:
        flag = 1
    
    else:
        X[n-1] = T[n-1,n]/T[n-1,n-1] 

        for i in range(n-2,-1,-1):
            # sum in row i
            s = 0
            for j in range(i+1, n):
                s += T[i,j]*X[j]
            X[i] = (T[i,n] - s)/T[i,i]
    
    return X, flag
```

```python
T, err = gaussian_partial(A,b)
X, err = backsubstitution(T)
```
