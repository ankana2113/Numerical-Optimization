# Find Determinant of a Matrix

$A = \begin{bmatrix}a & b \\ c & d \end{bmatrix}$

$det(A) = ad - bc$

```python
import numpy as np
```

```python
# find determinant of 2x2 matrix
A = np.array([[2,3],[1,2]])

A[0,0]*A[1,1] - A[1,0]*A[0,1]
```

If the dimension >= 2, we use the recursion to find the determinant. 

Steps to find determinant:

- Base case: when the determinant = 2, we use the method discussed above
- Else, we traverse the first row and find determinant correspoding to the element we traverse

```python
def findDet(A, total = 0):
    dim = A.shape[0]
    
    if dim == 2:
        det = A[0,0]*A[1,1] - A[1,0]*A[0,1]
        return det
    
    else:
        A1 = A.copy()

        # delete first row
        A1 = np.delete(A1,0,0)
        cols = A.shape[1]

        for i in range(cols):
            # delete column corresponding to position of element traversed
            A2 = np.delete(A1,i,1)
            total += (-1)**i * A[0,i] * findDet(A2)
            
    return total
```

```python
# check our result against Numpy.linalg.det()
A = np.random.normal(0,10,(3,3))
np.linalg.det(A)
```
