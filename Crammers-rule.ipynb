```python
import numpy as np
```

```python
def findDet(A, total = 0):
    dim = A.shape[0]

    if dim == 2:
        det = A[0,0]*A[1,1] - A[1,0]*A[0,1]
        return det
    
    else: 
        A1 = A.copy()

        # delete first row
        A1 = np.delete(A1,0,0)
        cols = A.shape[1]

        for i in range(cols):
            # delete column corresponding to position of element traversed
            A2 = np.delete(A1,i,1)
            total += (-1)**i * A[0,i] * findDet(A2)
            
    return total
```

```python
def crammer(A, b):
    n = A.shape[0]
    X = np.zeros((n))
    
    D = findDet(A)
    
    for i in range(n):
        Ai = A.copy()
        Ai[:,i] = b
        Di = findDet(Ai)
        X[i] = Di/D
    
    return X
```

```python
A = np.array([[1,1,0,3],
            [2,1,-1,1],
            [3,-1,-1,2],
            [-1,2,3,-1]])

b = np.array([4,1,-3,4])

print(A)
print(b)
```

```python
crammer(A, b)
```

```python
np.linalg.solve(A,b)
```
